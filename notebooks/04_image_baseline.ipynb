```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, datasets, transforms
from torch.utils.data import DataLoader

# Data
root = "../image_data"
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

train_ds = datasets.ImageFolder(f"{root}/train", transform=transform)
val_ds = datasets.ImageFolder(f"{root}/val", transform=transform)

train_dl = DataLoader(train_ds, batch_size=8, shuffle=True)
val_dl = DataLoader(val_ds, batch_size=8)

# Model
device = "cuda" if torch.cuda.is_available() else "cpu"
model = models.resnet18(pretrained=True)
for param in model.parameters():  # freeze backbone
    param.requires_grad = False
model.fc = nn.Linear(model.fc.in_features, 2)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=1e-3)

# Train 1 epoch (sanity check)
for imgs, labels in train_dl:
    imgs, labels = imgs.to(device), labels.to(device)
    out = model(imgs)
    loss = criterion(out, labels)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    break

print("Sanity check loss:", loss.item())
```
