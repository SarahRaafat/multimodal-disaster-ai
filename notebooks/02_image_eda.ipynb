```python
import matplotlib.pyplot as plt
import torchvision
from torchvision import datasets, transforms
import random

root = "../image_data"
transform = transforms.ToTensor()
train_ds = datasets.ImageFolder(f"{root}/train", transform=transform)

# Show class balance
labels = [train_ds.classes[i] for i in train_ds.targets]
plt.hist(labels)
plt.title("Class Balance (Train)")
plt.show()

# Show 12 random images
fig, axes = plt.subplots(3, 4, figsize=(10, 8))
for ax in axes.ravel():
    idx = random.randint(0, len(train_ds)-1)
    img, label = train_ds[idx]
    ax.imshow(img.permute(1,2,0))
    ax.set_title(train_ds.classes[label])
    ax.axis("off")
plt.show()
```
