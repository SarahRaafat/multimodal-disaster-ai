```python
import pandas as pd
import matplotlib.pyplot as plt

train = pd.read_csv("../text_data/train.csv")

# Class counts
counts = train['label'].value_counts()
counts.plot(kind='bar', title="Class Balance (Train)")
plt.show()

# Tweet length distribution
train['len'] = train['text'].str.len()
train['len'].hist(bins=50)
plt.title("Tweet Length Distribution")
plt.show()

# Show 5 random examples per class
for label in ['disaster', 'non_disaster']:
    print(f"\n=== {label.upper()} EXAMPLES ===")
    samples = train[train['label'] == label].sample(5, random_state=42)
    for t in samples['text']:
        print("-", t)
```
